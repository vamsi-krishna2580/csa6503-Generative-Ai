## STEP 1 — Import Required Libraries

This cell imports pandas and NumPy for loading and handling the wildlife dataset.


```python
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

## STEP 2 — Load the Wildlife Dataset

This cell loads the 250-record wildlife observation CSV into a pandas DataFrame.


```python
import pandas as pd

df = pd.read_csv("dataset/wildlife_observations.csv")

print("Dataset loaded successfully")
print("Number of records:", len(df))
print("Number of species:", df["common_name"].nunique())

df.head()
```

### Dataset Source and Citation

**Dataset source:** Synthetic citizen-science wildlife observation dataset created for educational benchmarking. The common-name/scientific-name pairings are based on publicly available wildlife taxonomy information; the individual observation descriptions are synthetic and not real sighting records. The dataset is used only for comparing semantic search with FAISS and ChromaDB.


## STEP 3 — Generate Text Embeddings

This cell loads all-MiniLM-L6-v2 and converts the 250 wildlife descriptions into 384-dimensional embeddings.


```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(
    df["description"].tolist(),
    show_progress_bar=True
)

print("Embeddings generated successfully")
print("Number of embeddings:", len(embeddings))
print("Embedding dimension:", embeddings.shape[1])
```

## STEP 3.1 — Verify Embeddings

This cell checks the first embedding and confirms the embedding matrix has the expected shape.


```python
print("First embedding:")
print(embeddings[0][:10])

print("\nEmbedding shape:")
print(embeddings.shape)
```

## STEP 4 — Build FAISS Index

This cell creates the FAISS vector index and measures the time required to index all 250 embeddings.


```python
import faiss
import time

# Convert embeddings to float32 for FAISS
embeddings_faiss = np.array(embeddings).astype("float32")

# Create FAISS index using L2 distance
index = faiss.IndexFlatL2(embeddings_faiss.shape[1])

# Measure indexing time
start_time = time.perf_counter()

index.add(embeddings_faiss)

faiss_indexing_time = time.perf_counter() - start_time

print("FAISS index created successfully")
print("Number of vectors:", index.ntotal)
print(f"FAISS indexing time: {faiss_indexing_time:.6f} seconds")
```

## STEP 4.1 — Save FAISS Index

This cell saves the FAISS index to the results folder for later use.


```python
faiss.write_index(index, "results/wildlife_faiss.index")

print("FAISS index saved successfully")
```

## STEP 5 — Build ChromaDB Collection

This cell creates the ChromaDB collection, stores the same embeddings and metadata, and measures indexing time.


```python
import chromadb
import time

# Create ChromaDB client
chroma_client = chromadb.PersistentClient(path="results/chroma_db")

# Create collection
collection = chroma_client.get_or_create_collection(
    name="wildlife_sightings"
)

# Prepare data
documents = df["description"].tolist()
ids = df["id"].astype(str).tolist()
embeddings_chroma = np.array(embeddings).tolist()

# Measure indexing time
start_time = time.perf_counter()

collection.add(
    ids=ids,
    documents=documents,
    embeddings=embeddings_chroma,
    metadatas=df[["common_name", "scientific_name"]].to_dict("records")
)

chroma_indexing_time = time.perf_counter() - start_time

print("ChromaDB collection created successfully")
print("Number of records:", collection.count())
print(f"ChromaDB indexing time: {chroma_indexing_time:.6f} seconds")
```

## STEP 5.1 — Verify ChromaDB

This cell verifies that ChromaDB contains all 250 wildlife records.


```python
print("ChromaDB record count:", collection.count())
```

## STEP 5.2 — Measure Storage Footprint

This cell measures the disk space used by the FAISS index and ChromaDB database.


```python
import os

faiss_size_bytes = os.path.getsize("results/wildlife_faiss.index")

chroma_size_bytes = 0
for root, dirs, files in os.walk("results/chroma_db"):
    for file in files:
        chroma_size_bytes += os.path.getsize(os.path.join(root, file))

print(f"FAISS index size: {faiss_size_bytes / 1024:.2f} KB")
print(f"ChromaDB storage size: {chroma_size_bytes / (1024 * 1024):.2f} MB")
```

## STEP 6 — Define Semantic Search Queries

This cell defines the 10 semantic-search queries required for benchmarking both databases.


```python
queries = [
    "A large striped predator seen near a forest at dusk",
    "A herd of large grey animals crossing a village road",
    "A colorful bird with a long tail near farmland",
    "A wild cat moving through dense vegetation",
    "A nocturnal animal searching for food near a village",
    "A large herbivore drinking near a water source",
    "A small wild animal spotted in agricultural fields",
    "A snake observed near rocks and vegetation",
    "A group of monkeys near trees",
    "Panthera tigris tigris observed near woodland"
]

print("Number of queries:", len(queries))
```

## STEP 7 — Benchmark FAISS and ChromaDB

This cell runs the same queries against both databases, measures query latency, and collects the Top-5 results.


```python
import time

benchmark_results = []
all_results = {}

for query in queries:
    query_embedding = model.encode([query]).astype("float32")

    # ---------- FAISS ----------
    start = time.perf_counter()

    distances, indices = index.search(query_embedding, 5)

    faiss_time = time.perf_counter() - start

    faiss_results = df.iloc[indices[0]][
        ["id", "common_name", "scientific_name", "description"]
    ].copy()

    # ---------- ChromaDB ----------
    start = time.perf_counter()

    chroma_result = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=5,
        include=["documents", "metadatas"]
    )

    chroma_time = time.perf_counter() - start

    chroma_results = pd.DataFrame({
        "id": chroma_result["ids"][0],
        "common_name": [
            m["common_name"] for m in chroma_result["metadatas"][0]
        ],
        "scientific_name": [
            m["scientific_name"] for m in chroma_result["metadatas"][0]
        ],
        "description": chroma_result["documents"][0]
    })

    all_results[query] = {
        "faiss": faiss_results,
        "chroma": chroma_results
    }

    benchmark_results.append({
        "query": query,
        "faiss_latency": faiss_time,
        "chroma_latency": chroma_time
    })

benchmark_df = pd.DataFrame(benchmark_results)

print("Benchmark completed successfully")
benchmark_df
```

## STEP 7.1 — Calculate Average Query Latency

This cell calculates the average search time for FAISS and ChromaDB across all 10 queries.


```python
avg_faiss = benchmark_df["faiss_latency"].mean()
avg_chroma = benchmark_df["chroma_latency"].mean()

print(f"FAISS average query latency: {avg_faiss:.6f} seconds")
print(f"ChromaDB average query latency: {avg_chroma:.6f} seconds")
```

## STEP 8 — Compare Top-5 Results

This cell displays the Top-5 results from FAISS and ChromaDB for the first three queries.


```python
for i, query in enumerate(queries[:3], 1):
    print("=" * 80)
    print(f"QUERY {i}: {query}")
    
    print("\nFAISS TOP-5:")
    display(all_results[query]["faiss"])
    
    print("\nCHROMADB TOP-5:")
    display(all_results[query]["chroma"])
```

## STEP 9 — Test the Special Case

This cell tests the scientific-name query for Panthera tigris tigris against both databases.


```python
special_query = queries[9]

print("SPECIAL CASE QUERY:")
print(special_query)

print("\nFAISS TOP-5:")
display(all_results[special_query]["faiss"])

print("\nCHROMADB TOP-5:")
display(all_results[special_query]["chroma"])
```

## STEP 10 — Define Expected Species

This cell defines the expected species for each semantic query so Top-5 relevance can be measured.


```python
expected_species = [
    "Bengal Tiger",
    "Asian Elephant",
    "Indian Peafowl",
    "Indian Leopard",
    "Sloth Bear",
    "Asian Elephant",
    "Indian Hare",
    "Indian Cobra",
    "Rhesus Macaque",
    "Bengal Tiger"
]
```

## STEP 11 — Calculate Top-5 Relevance

This cell counts how many of the Top-5 results match the expected species for each query.


```python
relevance_results = []

for query, expected in zip(queries, expected_species):

    faiss_top5 = all_results[query]["faiss"]
    chroma_top5 = all_results[query]["chroma"]

    faiss_relevant = sum(
        faiss_top5["common_name"].str.lower() == expected.lower()
    )

    # Chroma metadata needs to be queried separately
    chroma_check = collection.query(
        query_embeddings=model.encode([query]).tolist(),
        n_results=5,
        include=["documents", "metadatas"]
    )

    chroma_relevant = sum(
        m["common_name"].lower() == expected.lower()
        for m in chroma_check["metadatas"][0]
    )

    relevance_results.append({
        "query": query,
        "expected_species": expected,
        "FAISS_relevant_top5": faiss_relevant,
        "ChromaDB_relevant_top5": chroma_relevant
    })

relevance_df = pd.DataFrame(relevance_results)

relevance_df
```

## STEP 12 — Create Final Benchmark Table

This cell combines the dataset size, embedding size, indexing time, query latency, and relevance into the final comparison table.


```python
final_benchmark = pd.DataFrame({
    "Metric": [
        "Number of records",
        "Embedding dimension",
        "Indexing time (seconds)",
        "Average query latency (seconds)",
        "Average relevant results in Top-5",
        "Storage footprint"
    ],
    "FAISS": [
        len(df),
        embeddings.shape[1],
        faiss_indexing_time,
        avg_faiss,
        relevance_df["FAISS_relevant_top5"].mean(),
        f"{faiss_size_bytes / 1024:.2f} KB"
    ],
    "ChromaDB": [
        len(df),
        embeddings.shape[1],
        chroma_indexing_time,
        avg_chroma,
        relevance_df["ChromaDB_relevant_top5"].mean(),
        f"{chroma_size_bytes / (1024 * 1024):.2f} MB"
    ]
})

final_benchmark
```

## STEP 13 — Save Benchmark Results

This cell saves the latency, relevance, and final benchmark tables as CSV files in the results folder.


```python
benchmark_df.to_csv("results/query_latency.csv", index=False)
relevance_df.to_csv("results/relevance_results.csv", index=False)
final_benchmark.to_csv("results/final_benchmark.csv", index=False)

print("All benchmark results saved.")
```
